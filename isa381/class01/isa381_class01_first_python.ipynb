## Cell 2: Look at the data from class

This is the exact table from our East-vs-West exercise.

The first line brings in **pandas**, Python's tool for working with tables (think of it as Excel inside Python). The rest builds the table and shows it.

```python
import pandas as pd

df = pd.DataFrame({
    "region":          ["East", "East", "West", "West"],
    "month":           ["Jan",  "Feb",  "Jan",  "Feb"],
    "avg_order_value": [80.0,   40.0,   55.0,   56.0],
    "orders":          [10,     1000,   500,    520],
})

df
```

## Cell 3: Answer 1, the simple average

One way to answer "which region had the higher average order value?" is to average each region's two monthly numbers. In pandas, that reads almost like English: *group the table by region, take the mean of avg_order_value*.

```python
df.groupby("region")["avg_order_value"].mean()
```

By this answer, **East wins** ($60.00 vs $55.50).

But look at the table in Cell 2 again. East's $80 month had only **10 orders**. Its $40 month had **1,000**. Should those two months really count equally?

## Cell 4: Answer 2, the average across all orders

The other way to answer the question: across *all* the orders a region received, what was the average order value? That is total dollars divided by total orders.

```python
df["dollars"] = df["avg_order_value"] * df["orders"]   # dollars each month brought in

totals = df.groupby("region")[["dollars", "orders"]].sum()
totals["dollars"] / totals["orders"]
```

The answer **flipped**: now West wins ($55.50 vs $40.40), because East's impressive $80 month barely had any orders behind it.

Both computations ran without errors. Only one of them answers the question the business actually cares about. Telling them apart did not require being a programmer; it required reading a few lines of Python and asking one skeptical question. That is the skill this course builds.
